```python
import pandas as pd

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/covtype/covtype.data.gz"

colunas = [
    "Elevation",
    "Aspect",
    "Slope",
    "Horizontal_Distance_To_Hydrology",
    "Vertical_Distance_To_Hydrology",
    "Horizontal_Distance_To_Roadways",
    "Hillshade_9am",
    "Hillshade_Noon",
    "Hillshade_3pm",
    "Horizontal_Distance_To_Fire_Points"
]

colunas += [f"Wilderness_Area_{i}" for i in range(1, 5)]
colunas += [f"Soil_Type_{i}" for i in range(1, 41)]
colunas += ["Cover_Type"]

df = pd.read_csv(
    url,
    compression="gzip",
    header=None,
    names=colunas
)

df.head()
```

#### 4.3) **(20%)** Análise 3 - Correlação entre variáveis

O aluno deve apresentar 3 análises de correlação entre variáveis do conjunto de dados trabalhado. Exemplo: Em um conjunto de dados com as informações de temperatura e ocorrência de incêndios, eu gostaria de saber a incidência de correlação entre as duas variáveis.



Correlação 1 — Elevação × distância até estradas:

```python
df["Elevation"].corr(
    df["Horizontal_Distance_To_Roadways"]
)
```

Correlação 2 — Elevação × distância até pontos de incêndio:

```python
df["Elevation"].corr(
    df["Horizontal_Distance_To_Fire_Points"]
)
```

Correlação 3 — Distância até estradas × distância até pontos de incêndio:

```python
df["Horizontal_Distance_To_Roadways"].corr(
    df["Horizontal_Distance_To_Fire_Points"]
)
```

Matriz de correlação:

```python
variaveis_correlacao = [
    "Elevation",
    "Horizontal_Distance_To_Roadways",
    "Horizontal_Distance_To_Fire_Points"
]

correlacao = df[variaveis_correlacao].corr()

correlacao
```

```python
plt.figure(figsize=(8, 6))

plt.imshow(correlacao, cmap="coolwarm", vmin=-1, vmax=1)

plt.colorbar(label="Correlação")

plt.xticks(
    range(len(correlacao.columns)),
    correlacao.columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(correlacao.columns)),
    correlacao.columns
)

plt.title("Matriz de Correlação")

plt.tight_layout()
plt.show()
```

### Interpretação das correlações

A análise de correlação apresentou valores positivos para as três relações analisadas.

A correlação entre Elevation e Horizontal_Distance_To_Roadways foi de 0,365559, indicando uma relação positiva fraca a moderada. Isso significa que existe uma tendência de maiores valores de elevação estarem associados a maiores distâncias até as estradas, embora essa relação não seja forte.

A correlação entre Elevation e Horizontal_Distance_To_Fire_Points foi de 0,148022, indicando uma relação positiva fraca. Portanto, a elevação apresenta pouca associação linear com a distância até os pontos de incêndio.

Já a correlação entre Horizontal_Distance_To_Roadways e Horizontal_Distance_To_Fire_Points foi de 0,331580, indicando também uma relação positiva fraca a moderada.

De forma geral, os resultados mostram que existe associação positiva entre as variáveis analisadas, mas nenhuma das correlações apresentou valor suficientemente alto para indicar uma relação linear forte.
